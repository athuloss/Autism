```python
import opendatasets as od
od.download('https://www.kaggle.com/datasets/mathchi/diabetes-data-set')
```

```python
import pandas as pd

data = pd.read_csv('/content/diabetes-data-set/diabetes.csv')
data=data.drop(columns=['Pregnancies'])
print(data.to_string())
```

```python
cols_with_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
data[cols_with_zeros] = data[cols_with_zeros].replace(0, pd.NA)
data[cols_with_zeros] = data[cols_with_zeros].fillna(data[cols_with_zeros].median())
print(data.head())
print(data.info())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X=data.drop(columns=['Outcome'])
y=data['Outcome']

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model = LogisticRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
print(y_pred)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
counts = data['Outcome'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(['Non-Diabetic', 'Diabetic'], counts, color=['skyblue', 'salmon'])
plt.title('Number of Diabetic vs Non-Diabetic Patients')
plt.ylabel('Count')
plt.show()
```
